```python

import pandas as pd
from google.colab import files
import io
import zipfile


uploaded = files.upload()

filename = next(iter(uploaded))
file_bytes = uploaded[filename]


if filename.lower().endswith(".zip"):
    with zipfile.ZipFile(io.BytesIO(file_bytes)) as z:
        csv_files = [f for f in z.namelist()
                     if f.lower().endswith(".csv")
                     and not f.startswith("__MACOSX/")]

        if not csv_files:
            raise ValueError("No CSV file found inside ZIP.")

        print("CSV selected:", csv_files[0])
        df = pd.read_csv(z.open(csv_files[0]))

elif filename.lower().endswith(".csv"):
    df = pd.read_csv(io.BytesIO(file_bytes))

else:
    raise ValueError("Please upload a CSV or ZIP file.")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# 1. Remove duplicate records
df = df.drop_duplicates().copy()

# 2. Remove identifier and constant columns
drop_cols = [
    col for col in [
        "EmployeeNumber",
        "EmployeeCount",
        "Over18",
        "StandardHours"
    ] if col in df.columns
]

df = df.drop(columns=drop_cols)

# 3. Separate target and features
if "Attrition" not in df.columns:
    raise ValueError("Target column 'Attrition' was not found.")

y = df["Attrition"].astype(str).str.strip().map(
    {"Yes": 1, "No": 0}
)

if y.isna().any():
    raise ValueError("Unexpected values found in Attrition column.")

X = df.drop(columns=["Attrition"])

# 4. Identify numeric and categorical features
numeric_features = X.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X.select_dtypes(
    exclude=["number"]
).columns.tolist()

# 5. Preprocessing pipelines
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

# 6. Train-test split with stratification
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 7. Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed successfully!")
print("Training samples:", X_train_processed.shape[0])
print("Testing samples:", X_test_processed.shape[0])
print("Processed features:", X_train_processed.shape[1])
print("\nTraining target distribution:")
print(y_train.value_counts())
print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
from scipy.sparse import issparse

# GaussianNB and LDA require dense arrays
X_train_dense = (
    X_train_processed.toarray()
    if issparse(X_train_processed)
    else X_train_processed
)

X_test_dense = (
    X_test_processed.toarray()
    if issparse(X_test_processed)
    else X_test_processed
)

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000, class_weight="balanced", random_state=42
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),
    "Gaussian Naive Bayes": GaussianNB(),
    "Linear Discriminant Analysis": LinearDiscriminantAnalysis(),
    "Neural Network": MLPClassifier(
        hidden_layer_sizes=(64, 32),
        max_iter=300,
        early_stopping=True,
        random_state=42
    )
}

results = []
trained_models = {}

for name, model in models.items():
    try:
        # Dense input is needed for GaussianNB and LDA
        if name in [
            "Gaussian Naive Bayes",
            "Linear Discriminant Analysis"
        ]:
            model.fit(X_train_dense, y_train)
            predictions = model.predict(X_test_dense)
            probabilities = model.predict_proba(X_test_dense)[:, 1]
        else:
            model.fit(X_train_processed, y_train)
            predictions = model.predict(X_test_processed)
            probabilities = model.predict_proba(X_test_processed)[:, 1]

        results.append({
            "Model": name,
            "Accuracy": accuracy_score(y_test, predictions),
            "Precision": precision_score(
                y_test, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_test, predictions, zero_division=0
            ),
            "F1 Score": f1_score(
                y_test, predictions, zero_division=0
            ),
            "ROC-AUC": roc_auc_score(y_test, probabilities)
        })

        trained_models[name] = model
        print(f"Completed: {name}")

    except Exception as e:
        print(f"Error in {name}: {e}")

results_df = pd.DataFrame(results).sort_values(
    by="F1 Score", ascending=False
).reset_index(drop=True)

display(results_df.round(4))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Display model comparison
display(
    results_df.style.format({
        "Accuracy": "{:.2%}",
        "Precision": "{:.2%}",
        "Recall": "{:.2%}",
        "F1 Score": "{:.2%}",
        "ROC-AUC": "{:.4f}"
    })
)

# Visualize F1 Score
plt.figure(figsize=(10, 5))
sns.barplot(data=results_df, x="F1 Score", y="Model")
plt.title("Model Comparison - F1 Score")
plt.xlim(0, 1)
plt.tight_layout()
plt.show()

# Select best model based on F1 Score
best_model_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_model_name]

print("Best model by F1 Score:", best_model_name)
```

```python

%%writefile app.py
import streamlit as st
import pandas as pd
import plotly.express as px

st.set_page_config(
    page_title="Employee Attrition Analytics",
    page_icon="📊",
    layout="wide"
)

st.title("Employee Attrition Analytics Platform")
st.caption("HR Workforce Insights and Model Performance")

uploaded = st.sidebar.file_uploader(
    "Upload HR Attrition CSV",
    type=["csv"]
)

if uploaded is not None:
    df = pd.read_csv(uploaded)
else:
    st.info("Upload your IBM HR Attrition CSV file to view the dashboard.")
    st.stop()

required = {"Attrition", "Department"}
if not required.issubset(df.columns):
    st.error("CSV must contain Attrition and Department columns.")
    st.stop()

df = df.drop_duplicates().copy()
df["Attrition"] = df["Attrition"].astype(str).str.strip()

total = len(df)
left = int((df["Attrition"] == "Yes").sum())
retained = int((df["Attrition"] == "No").sum())
attrition_rate = left / total * 100 if total else 0
retention_rate = retained / total * 100 if total else 0

c1, c2, c3, c4 = st.columns(4)
c1.metric("Total Employees", f"{total:,}")
c2.metric("Employees Left", f"{left:,}")
c3.metric("Attrition Rate", f"{attrition_rate:.2f}%")
c4.metric("Retention Rate", f"{retention_rate:.2f}%")

st.subheader("Department-wise Attrition")

dept = df.groupby("Department").agg(
    Total=("Attrition", "size"),
    Left=("Attrition", lambda x: (x == "Yes").sum())
).reset_index()

dept["Attrition Rate (%)"] = (
    dept["Left"] / dept["Total"] * 100
).round(2)

fig = px.bar(
    dept,
    x="Department",
    y="Attrition Rate (%)",
    color="Department",
    hover_data=["Total", "Left"],
    title="Attrition Rate by Department"
)
st.plotly_chart(fig, use_container_width=True)

left_col, right_col = st.columns(2)

with left_col:
    st.subheader("Attrition Distribution")
    counts = df["Attrition"].value_counts().rename_axis(
        "Attrition"
    ).reset_index(name="Employees")
    fig2 = px.pie(
        counts,
        names="Attrition",
        values="Employees",
        hole=0.4
    )
    st.plotly_chart(fig2, use_container_width=True)

with right_col:
    st.subheader("Workforce Demographics")
    if "Gender" in df.columns:
        gender = df["Gender"].value_counts().rename_axis(
            "Gender"
        ).reset_index(name="Employees")
        fig3 = px.bar(
            gender,
            x="Gender",
            y="Employees",
            title="Employees by Gender"
        )
        st.plotly_chart(fig3, use_container_width=True)

st.subheader("Department Summary")
st.dataframe(dept, use_container_width=True)

st.subheader("Employee Records")
st.dataframe(df.head(100), use_container_width=True)

st.download_button(
    "Download Department Summary",
    data=dept.to_csv(index=False).encode("utf-8"),
    file_name="department_attrition_summary.csv",
    mime="text/csv"
)
```

```python

%%writefile app.py
import streamlit as st
import pandas as pd
import plotly.express as px
import os
import glob

st.set_page_config(
    page_title="Employee Attrition Analytics",
    page_icon="📊",
    layout="wide"
)

st.title("Employee Attrition Analytics Platform")
st.caption("HR Workforce Insights and Analytics")

# Find the CSV already uploaded to Colab
csv_files = [
    f for f in glob.glob("/content/*.csv")
    if os.path.isfile(f)
]

if not csv_files:
    st.error(
        "CSV file not found in Colab. "
        "Upload your HR dataset in a Colab cell first."
    )
    st.stop()

# Prefer the IBM HR Attrition dataset if multiple CSV files exist
preferred = [
    f for f in csv_files
    if "attrition" in os.path.basename(f).lower()
]

csv_path = preferred[0] if preferred else csv_files[0]

try:
    df = pd.read_csv(csv_path)
except Exception as e:
    st.error(f"Could not read CSV: {e}")
    st.stop()

if not {"Attrition", "Department"}.issubset(df.columns):
    st.error(
        "Incorrect dataset. Please use the IBM HR Attrition CSV."
    )
    st.stop()

df = df.drop_duplicates().copy()
df["Attrition"] = df["Attrition"].astype(str).str.strip()

total = len(df)
left = int((df["Attrition"] == "Yes").sum())
retained = int((df["Attrition"] == "No").sum())

attrition_rate = left / total * 100 if total else 0
retention_rate = retained / total * 100 if total else 0

st.success(f"Dataset loaded: {os.path.basename(csv_path)}")

# KPI cards
c1, c2, c3, c4 = st.columns(4)
c1.metric("Total Employees", f"{total:,}")
c2.metric("Employees Left", f"{left:,}")
c3.metric("Attrition Rate", f"{attrition_rate:.2f}%")
c4.metric("Retention Rate", f"{retention_rate:.2f}%")

# Department analytics
st.subheader("Department-wise Attrition")

dept = df.groupby("Department").agg(
    Total_Employees=("Attrition", "size"),
    Employees_Left=("Attrition", lambda x: (x == "Yes").sum())
).reset_index()

dept["Attrition_Rate_%"] = (
    dept["Employees_Left"] / dept["Total_Employees"] * 100
).round(2)

fig = px.bar(
    dept,
    x="Department",
    y="Attrition_Rate_%",
    color="Department",
    hover_data=["Total_Employees", "Employees_Left"],
    title="Attrition Rate by Department"
)
st.plotly_chart(fig, use_container_width=True)

# Attrition distribution
st.subheader("Attrition Distribution")

counts = df["Attrition"].value_counts().rename_axis(
    "Attrition"
).reset_index(name="Employees")

fig2 = px.pie(
    counts,
    names="Attrition",
    values="Employees",
    hole=0.4
)
st.plotly_chart(fig2, use_container_width=True)

# Demographics
st.subheader("Workforce Demographics")

if "Gender" in df.columns:
    gender = df["Gender"].value_counts().rename_axis(
        "Gender"
    ).reset_index(name="Employees")

    st.plotly_chart(
        px.bar(gender, x="Gender", y="Employees"),
        use_container_width=True
    )

if "Age" in df.columns:
    st.plotly_chart(
        px.histogram(df, x="Age", color="Attrition",
                     title="Age Distribution by Attrition"),
        use_container_width=True
    )

st.subheader("Department Summary")
st.dataframe(dept, use_container_width=True)

st.subheader("Employee Data Preview")
st.dataframe(df.head(100), use_container_width=True)

st.download_button(
    "Download Department Summary",
    data=dept.to_csv(index=False).encode("utf-8"),
    file_name="department_attrition_summary.csv",
    mime="text/csv"
)
```
